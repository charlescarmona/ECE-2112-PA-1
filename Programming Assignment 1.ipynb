# A. Word Rotation Problem - rotates the letters of the word

```python
def rotate_word(text): #defines the name of the function and variable
    #Cuts the first character of the word and places it at the end of the word
    return text[1:] + text[0]
```

```python
rotate_word("python") #This tests the function
```

```python
#This prints the output for each variable
print(rotate_word("python"))
print(rotate_word("logic"))
print(rotate_word("Code"))
print(rotate_word("A"))
```

# B. Username Builder Problem - builds a username from the given names

```python
def make_username(first_name, last_name): #defines the name of the function and variables
    #This makes the letters of the first name and last name in lowercase and removes all the spaces
    first_name = first_name.lower().replace(" ","")
    last_name = last_name.lower().replace(" ","")
    return first_name + "." + last_name
```

```python
print(make_username("Ada", "Lovelace")) #this tests the function
```

```python
#prints the output for each variable under the function
print(make_username("Ada", "Lovelace"))
print(make_username("Alan", "Turing"))
print(make_username("Ana Maria", "De Leon"))
```

# C. Bookend Swap Problem - swaps first and last elements

```python
def swap_bookends(items): #defines the name of the function and variable
    #this separates the elements into sections, being first, middle and last
    first, *middle, last = items
    #this swaps the first and last elements
    return [last] + middle + [first]
```

```python
swap_bookends ([1,2,3,4,5,6]) #this tests the function
```

```python
#this prints the output for each variable
print(swap_bookends ([1,2,3,4,5,6]))
print(swap_bookends (["red", "green", "blue"]))
print(swap_bookends ([8,3]))
```
